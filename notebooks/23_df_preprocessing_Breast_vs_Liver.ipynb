# Genomic Dataframe Breast vs Liver

```python
import pandas as pd

# ── Paths ─────────────────────────────────────────────────────────────────────
gene_expression_path = "/mnt/bulk-saturn/maralampert/genhist/dataframes/tcga_RNASeq.csv"
output_path          = "/mnt/bulk-saturn/maralampert/genhist/dataframes/tcga_RNASeq_BRCA_LIHC.csv"

# ── TSS-code → cohort mapping (from TCGA BCR) ─────────────────────────────────
BRCA_TSS = {
    '3C','4H','5L','5T','A1','A2','A7','A8','AC','AN','AO','AQ','AR',
    'B6','BH','C8','D8','E2','E9','EW','GI','GM','HN','JL','LD','LL',
    'OK','OL','PE','PL','S3','UL','UU','V7','W8','WT','XX','Z7'
}
LIHC_TSS = {
    'CC','DD','ED','EP','ES','FV','G3','K7','KR','LN','MI','RC','XR','ZH'
}

# ── Load & annotate cohort ────────────────────────────────────────────────────
df = pd.read_csv(gene_expression_path, index_col=0)
tss = df["Patient_ID"].str.split("-").str[1]

df["cohort"] = "other"
df.loc[tss.isin(BRCA_TSS), "cohort"] = "TCGA-BRCA"
df.loc[tss.isin(LIHC_TSS), "cohort"] = "TCGA-LIHC"

# ── Filter ────────────────────────────────────────────────────────────────────
df_filtered = df[df["cohort"].isin(["TCGA-BRCA", "TCGA-LIHC"])].copy()

print(f"Total samples:  {len(df):,}")
print(f"BRCA:           {(df_filtered['cohort'] == 'TCGA-BRCA').sum():,}")
print(f"LIHC:           {(df_filtered['cohort'] == 'TCGA-LIHC').sum():,}")
print(f"Output shape:   {df_filtered.shape}")

# ── Save ──────────────────────────────────────────────────────────────────────
df_filtered.to_csv(output_path)
print(f"\nSaved → {output_path}")
```
