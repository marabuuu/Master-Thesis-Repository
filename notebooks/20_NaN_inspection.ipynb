# NaN Inspection

```python
df_path = '../../dataframes/brca_gene_expression_with_subtypes.csv'
```

```python
# read in df
import pandas as pd
df = pd.read_csv(df_path, index_col=0)

# check for NaN values
nan_counts = df.isna().sum()
print(nan_counts[nan_counts > 0])
```

The dataframe has zero missing entries; every gene expression value and subtype field is filled.
