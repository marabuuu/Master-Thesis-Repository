# Opening zipped image tiles

This notebook serves as an investigation of the zip files used for image tiles.

```python
import os, zipfile
from collections import Counter
from PIL import Image
```

```python
data_dir = "/mnt/bulk-saturn/maralampert/genhist/data/TCGA"
```

```python
# please open a specific zip file and print some filenames of png files so I understand its structure
zip_path = os.path.join(data_dir, "TCGA-3C-AALI-01Z-00-DX1.F6E9A5DF-D8FB-45CF-B4BD-C6B76294C291.b712aacbf58f47d612cb43986a2f97a21bcea971a4beb2070055f123940c1afd.zip")
with zipfile.ZipFile(zip_path, 'r') as zip_ref:
    zip_ref.extractall(data_dir)

    # Print some filenames of png files inside the zip
    png_files = [f for f in zip_ref.namelist() if f.endswith('.png')]
    print("Some PNG files inside the zip:")
    for f in png_files[:10]:  # print first 10 png files
        print(f)
```

```python
z = zipfile.ZipFile(os.path.join(data_dir, [f for f in os.listdir(data_dir) if f.lower().endswith('.zip')][0])); print(f"ZIP size: {os.path.getsize(z.filename)} bytes\nFiles: {len(z.namelist())}\nFirst file: {z.infolist()[0].filename} ({z.infolist()[0].file_size} bytes)")
```

```python
# print some filenames of the png files inside the zip so we understand its structure
png_files = [f for f in z.namelist() if f.lower().endswith('.png')]
print(f"Number of PNG files in ZIP: {len(png_files)}\nSome PNG files:\n" + "\n".join(png_files[:5]) + "\n...\n" + "\n".join(png_files[-5:]))    
```

```python
# please show the content of tiler_params.json
with z.open('tiler_params.json') as f:
    print(f.read().decode())
```

This prints the size of the zip file, the number of files inside it, and the name and size of the first file in the archive.

```python
# print out which different kinds of files are in one zip file and how many of each
file_types = [os.path.splitext(f)[1].lower() for f in z.namelist()]
file_type_counts = Counter(file_types)
for file_type, count in file_type_counts.items():
    print(f"{file_type}: {count}")  
```

```python
# pleae tell me about one png file the dimensions
with z.open([f for f in z.namelist() if f.lower().endswith('.png')][0]) as img_file:
    img = Image.open(img_file)
    print(f"Image size: {img.size} pixels")
```

We know now that we have in each zip file a `.json` and multiple thousand tiles as `.png` shaped 512x512.
