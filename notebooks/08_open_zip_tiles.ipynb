# Opening zipped image tiles

This notebook serves as an investigation of the zip files used for image tiles.

```python
import os, zipfile
from collections import Counter
from PIL import Image
```

```python
data_dir = "/data/horse/ws/mala059b-rna2wsi/data/TCGA/"
```

```python
z = zipfile.ZipFile(os.path.join(data_dir, [f for f in os.listdir(data_dir) if f.lower().endswith('.zip')][0])); print(f"ZIP size: {os.path.getsize(z.filename)} bytes\nFiles: {len(z.namelist())}\nFirst file: {z.infolist()[0].filename} ({z.infolist()[0].file_size} bytes)")
```

This prints the size of the zip file, the number of files inside it, and the name and size of the first file in the archive.

```python
# print out which different kinds of files are in one zip file and how many of each
file_types = [os.path.splitext(f)[1].lower() for f in z.namelist()]
file_type_counts = Counter(file_types)
for file_type, count in file_type_counts.items():
    print(f"{file_type}: {count}")  
```

```python
# pleae tell me about one png file the dimensions
with z.open([f for f in z.namelist() if f.lower().endswith('.png')][0]) as img_file:
    img = Image.open(img_file)
    print(f"Image size: {img.size} pixels")
```

We know now that we have in each zip file a `.json` and multiple thousand tiles as `.png` shaped 512x512.
