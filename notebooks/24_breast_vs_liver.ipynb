```python
breast_example = "/mnt/bulk-saturn/maralampert/genhist/data/BRCA-tumor-tiles-128/TCGA-3C-AALI-01Z-00-DX1.F6E9A5DF-D8FB-45CF-B4BD-C6B76294C291.b712aacbf58f47d612cb43986a2f97a21bcea971a4beb2070055f123940c1afd.zip"
liver_example = "/mnt/bulk-saturn/maralampert/genhist/data/Liver-tumor-tiles-128/TCGA-2Y-A9GS-01Z-00-DX1.75C5CEF1-6C8B-42FF-82F6-9A05621E9A80.e16e63fc68120526c3637a9d43cc8550280785025b3824441cc16a606d66fe46.zip"
```

```python
import zipfile, random
from PIL import Image
import matplotlib.pyplot as plt
import io

def sample_pngs(zip_path, n=3, seed=42):
    with zipfile.ZipFile(zip_path) as zf:
        names = [m for m in zf.namelist() if m.lower().endswith(".png")]
        random.seed(seed)
        chosen = random.sample(names, min(n, len(names)))
        imgs = []
        for name in chosen:
            imgs.append(Image.open(io.BytesIO(zf.read(name))).convert("RGB"))
    return imgs

breast_imgs = sample_pngs(breast_example)
liver_imgs  = sample_pngs(liver_example)

fig, axes = plt.subplots(2, 3, figsize=(10, 7))
for i, img in enumerate(breast_imgs):
    axes[0, i].imshow(img); axes[0, i].axis("off")
    axes[0, i].set_title(f"Breast {i+1}", fontsize=9)
for i, img in enumerate(liver_imgs):
    axes[1, i].imshow(img); axes[1, i].axis("off")
    axes[1, i].set_title(f"Liver {i+1}", fontsize=9)
plt.tight_layout()
plt.show()
```
