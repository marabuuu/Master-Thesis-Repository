# Data Exploration

```python
import pandas as pd
import numpy as np
import plotly.express as px
import umap 
```

```python
# load csv file 
brca_df = pd.read_csv('../../data/brca_gene_expression_with_subtypes.csv')
```

## Plotting UMAP of all genes

```python
# create a UMAP using plotly by utilizing column Majority_Subtype_mRNA for coloring
reducer = umap.UMAP(random_state=42)
embedding = reducer.fit_transform(brca_df.drop(columns=['Patient_ID', 'Majority_Subtype_mRNA']))
brca_df['UMAP1'] = embedding[:, 0]
brca_df['UMAP2'] = embedding[:, 1]
fig = px.scatter(brca_df, x='UMAP1', y='UMAP2', color='Majority_Subtype_mRNA',
                 title='UMAP of BRCA Gene Expression Colored by PAM50 Subtypes')
fig.show()
```

```python
# save the umap as an html file 
fig.write_html('../plots/umap_brca_all_subtypes.html')
```

## Plot UMAP of PAM50 genes 

```python
# read in txt file with list of genes
with open('../../data/pam50_gene_list.txt', 'r') as file:
    pam50_genes = [line.strip() for line in file.readlines()]
```

```python
# filter the columns of brca_df to only include pam50 genes
brca_pam50_df = brca_df[['Patient_ID', 'Majority_Subtype_mRNA'] + pam50_genes]
```

```python
# create a UMAP using plotly by utilizing column Majority_Subtype_mRNA for coloring
reducer = umap.UMAP(random_state=42)
embedding = reducer.fit_transform(brca_pam50_df.drop(columns=['Patient_ID', 'Majority_Subtype_mRNA']))
brca_pam50_df['UMAP1'] = embedding[:, 0]
brca_pam50_df['UMAP2'] = embedding[:, 1]
fig = px.scatter(brca_pam50_df, x='UMAP1', y='UMAP2', color='Majority_Subtype_mRNA',
                 title='UMAP of BRCA Gene Expression Colored by PAM50 Subtypes')
fig.show()
```

```python
# write the umap as an html file
fig.write_html('../plots/umap_brca_pam50_all_subtypes.html')
```
