# Z-score normalization of gene expression counts

```python
import pandas as pd 
import scanpy as sc  
import umap
import plotly.express as px
```

```python
brca_df_path = "../../data/brca_gene_expression_with_subtypes.csv"

brca_df = pd.read_csv(brca_df_path)
```

```python
# Preprocess data
expression_data = brca_df.drop(columns=['Patient_ID', 'Majority_Subtype_mRNA'])
scaled_X = sc.pp.scale(expression_data.values)  
```

```python
# UMAP with same parameters
reducer = umap.UMAP(random_state=42, n_neighbors=30, min_dist=0.2)
embedding = reducer.fit_transform(scaled_X)

# Create a dataframe with the UMAP coordinates
umap_df = pd.DataFrame({'UMAP1': embedding[:, 0], 'UMAP2': embedding[:, 1], 
                        'Majority_Subtype_mRNA': brca_df['Majority_Subtype_mRNA']})

# Plot using plotly
fig = px.scatter(umap_df, x='UMAP1', y='UMAP2', color='Majority_Subtype_mRNA',
                 title='UMAP of BRCA Gene Expression Colored by PAM50 Subtypes')
fig.show()
```

```python
# write the z score normalized data to a csv file 
z_score_df = pd.DataFrame(scaled_X, columns=expression_data.columns)
z_score_df['Patient_ID'] = brca_df['Patient_ID']
z_score_df.to_csv("../../data/brca_gene_expression_z_score_normalized.csv", index=False)
```
