# Check how many zip files are in a directory

```python
dir_path = "/mnt/bulk-saturn/maralampert/genhist/data/BRCA-tumor-tiles-all"
```

```python
# check how many zip files are in the dir 
import os
zip_files = [f for f in os.listdir(dir_path) if f.endswith('.zip')]
print(f"Number of zip files in the directory: {len(zip_files)}")
```
