# Inspect h5-files

```python
import h5py
import os
```

```python
foxa1_h5 = "/mnt/bulk-saturn/maralampert/genhist/experiments/20260306_FOXA1_finetuning/encoding/TCGA-C8-A12P.h5"
```

```python
gene_h5 = "/mnt/bulk-saturn/maralampert/genhist/experiments/20260316_joint_training/encoding_latents/TCGA-3C-AALI.h5"
```

```python
def inspect_h5(path):
    """Print dataset names, shapes and dtypes inside an HDF5 file."""
    print(f"Inspecting {path}")
    with h5py.File(path, "r") as f:
        def visitor(name, obj):
            if isinstance(obj, h5py.Dataset):
                print(f"  {name} : shape={obj.shape} dtype={obj.dtype}")
            elif isinstance(obj, h5py.Group):
                print(f"  {name}/ (group)")
        f.visititems(visitor)
    print("\n")

# run on the two paths defined above
for p in (foxa1_h5, gene_h5):
    inspect_h5(p)
```
