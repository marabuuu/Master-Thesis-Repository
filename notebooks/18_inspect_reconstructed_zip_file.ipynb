```python
#please open the zip file and inspect its contents. What files are inside?
import zipfile
with zipfile.ZipFile(reconstructed_zip_path, 'r') as zip_ref:
    file_list = zip_ref.namelist()
    print("Files inside the reconstructed zip:")
    for f in file_list:
        print(f)
```

```python
# please open one of the png files and display it
import matplotlib.pyplot as plt
from PIL import Image
import io

# Open the first PNG file in the zip
with zipfile.ZipFile(reconstructed_zip_path, 'r') as zip_ref:
    png_files = [f for f in zip_ref.namelist() if f.endswith('.png')]
    if png_files:
        first_png = png_files[0]
        print(f"Opening {first_png}...")
        with zip_ref.open(first_png) as file:
            img_data = file.read()
            img = Image.open(io.BytesIO(img_data))
            plt.imshow(img)
            plt.axis('off')
            plt.title(first_png)
            plt.show()
    else:
        print("No PNG files found in the zip.")
```
